```python
import pandas as pd
import scipy.stats as stats

df = pd.read_csv("/home/dell/ML-Learning/datasets/online_retail_II_cleaned.csv")
```

```python
sample = df["price"]
t_stat, p_value = stats.ttest_1samp(sample, 3)

print("T Statistic =", t_stat)
print("P Value =", p_value)
# note: this is a two-sided test (checks if mean is different from 3,
# not specifically greater than 3) - t_stat sign tells us the direction
```

```python
if p_value < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
    
```

```python
uk = df[df["country"] == "United Kingdom"]["quantity"]
non_uk = df[df["country"] != "United Kingdom"]["quantity"]

t, p = stats.ttest_ind(uk, non_uk, equal_var=False)

print(t)
print(p)
```

```python
if p < 0.05:
    print("Quantity differs significantly.")
else:
    print("No significant difference.")
```

## Observations

### Test 1: Average Price

- **Null Hypothesis (H₀):** The average price is equal to £3.
- **Alternative Hypothesis (H₁):** The average price is different from £3.
- **Significance Level:** 0.05

Since the p-value is almost 0 and is less than 0.05, we reject the null hypothesis. This means the average price is significantly different from £3. As the t-statistic is positive (77.1), the average price (£3.37) is actually higher than £3.
