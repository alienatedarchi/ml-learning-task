```python
import pandas as pd

customers = pd.read_csv("/home/dell/ML-Learning/datasets/customer_features.csv")
print(customers.shape)
customers.head()
```

```python
stats_cols = ["recency_days", "frequency", "monetary", "avg_order_value",
              "unique_products_bought", "customer_tenure_days"]

descriptive_stats = customers[stats_cols].describe().round(2)
descriptive_stats
```

```python
def build_stats_table(df, columns):
    rows = []
    for col in columns:
        data = df[col].dropna()
        rows.append({
            "feature": col,
            "mean": round(data.mean(), 2),
            "median": round(data.median(), 2),
            "mode": round(data.mode().iloc[0], 2) if not data.mode().empty else None,
            "std_dev": round(data.std(), 2),
            "min": round(data.min(), 2),
            "max": round(data.max(), 2),
            "range": round(data.max() - data.min(), 2),
            "skew": round(data.skew(), 2),   # >0 = right-skewed (long tail of high spenders)
        })
    return pd.DataFrame(rows)

stats_table = build_stats_table(customers, stats_cols)
stats_table
```

```python
percentile_table = customers[stats_cols].quantile([0.1, 0.25, 0.5, 0.75, 0.9, 0.99]).round(2)
percentile_table
```

```python
repeat_comparison = (
    customers.groupby("is_repeat_customer")[["monetary", "frequency", "avg_order_value", "recency_days"]]
    .mean()
    .round(2)
)
repeat_comparison
```
