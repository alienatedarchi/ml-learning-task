```python
# Simulation of a Coin Toss

import numpy as np

n = int(input("Enter number of tosses: "))

heads = 0
tails = 0

result = np.random.randint(0,2,n)

for i in range(n):

    if result[i] == 0:
        heads += 1
    else:
        tails += 1

print("Heads =", heads)
print("Tails =", tails)

print("Probability of Heads =", heads/n)
print("Probability of Tails =", tails/n)
```

```python
# Simulation of Rolling a Die

import numpy as np

n = int(input("Enter number of rolls: "))

count = [0]*6

roll = np.random.randint(1,7,n)

for i in range(n):

    if roll[i] == 1:
        count[0] += 1

    elif roll[i] == 2:
        count[1] += 1

    elif roll[i] == 3:
        count[2] += 1

    elif roll[i] == 4:
        count[3] += 1

    elif roll[i] == 5:
        count[4] += 1

    else:
        count[5] += 1

for i in range(6):
    print(i+1,"=",count[i])

num = int(input("Enter number whose probability you want: "))

print("Probability =",count[num-1]/n)
```

```python
# Simulation of Drawing a Card from a Deck

import numpy as np

n = int(input("Enter number of simulations: "))

cards = np.random.randint(1, 53, n)

ace = 0
king = 0
heart = 0

for i in range(n):

    if cards[i] == 1 or cards[i] == 14 or cards[i] == 27 or cards[i] == 40:
        ace += 1

    if cards[i] == 13 or cards[i] == 26 or cards[i] == 39 or cards[i] == 52:
        king += 1

    if cards[i] >= 1 and cards[i] <= 13:
        heart += 1

print("Number of Aces =", ace)
print("Number of Kings =", king)
print("Number of Hearts =", heart)

print("P(Ace) =", ace/n)
print("P(King) =", king/n)
print("P(Heart) =", heart/n)
```

```python
# Birthday Problem Simulation

import numpy as np

people = int(input("Enter number of people in a room: "))
simulations = int(input("Enter number of simulations: "))

count = 0

for i in range(simulations):

    birthdays = np.random.randint(1, 366, people)

    if len(set(birthdays)) < people:
        count += 1

print("Probability that two people share a birthday =", count/simulations)
```

```python
# Random Walk Simulation

import numpy as np

steps = int(input("Enter number of steps: "))

position = 0

move = np.random.randint(0, 2, steps)

for i in range(steps):

    if move[i] == 0:
        position = position - 1

    else:
        position = position + 1

print("Final Position =", position)

if position > 0:
    print("Walk ended on Positive Side")

elif position < 0:
    print("Walk ended on Negative Side")

else:
    print("Walk ended at Origin")
```

```python
# Marble Drawing Simulation

import numpy as np

red = int(input("Enter number of Red marbles: "))
blue = int(input("Enter number of Blue marbles: "))
green = int(input("Enter number of Green marbles: "))

simulations = int(input("Enter number of simulations: "))

red_count = 0
blue_count = 0
green_count = 0

total = red + blue + green

draw = np.random.randint(1, total + 1, simulations)

for i in range(simulations):

    if draw[i] <= red:
        red_count += 1

    elif draw[i] <= red + blue:
        blue_count += 1

    else:
        green_count += 1

print("Red Marbles Drawn =", red_count)
print("Blue Marbles Drawn =", blue_count)
print("Green Marbles Drawn =", green_count)

print("P(Red) =", red_count / simulations)
print("P(Blue) =", blue_count / simulations)
print("P(Green) =", green_count / simulations)
```
