```python
set_1 = {1, 2, 3, 4}
set_2 = {3, 4, 5, 6}
# Union of two sets
union_set = set_1.union(set_2)
print(union_set)  # Output: {1, 2, 3, 4, 5, 6}
# Intersection of two sets
intersection_set = set_1.intersection(set_2)
print(intersection_set)  # Output: {3, 4}
# Difference of two sets
difference_set = set_1.difference(set_2)
print(difference_set)  # Output: {1, 2}     
```

##### Question 2 - How do you check if a number is a prime factor of another number in Python?

```python
numb = int(input("Enter the number: "))
factor = int(input("Enter the factor to check: "))

is_prime = True

if factor <= 1:
    is_prime = False
else:
    for i in range(2, factor):
        if factor % i == 0:
            is_prime = False
            break


if is_prime and numb % factor == 0:
    print(f"{factor} is a prime factor of {numb}.")
else:
    print(f"{factor} is NOT a prime factor of {numb}.")
```

##### Question 5 - How do you find the longest consecutive sequence in an unsorted list?

```python

numbers = list(map(int, input("Enter numbers separated by spaces: ").split()))

num_set = set(numbers)

longest = 0

for num in num_set:
    # Check if it's the start of a sequence
    if num - 1 not in num_set:
        current = num
        length = 1

        while current + 1 in num_set:
            current += 1
            length += 1

        if length > longest:
            longest = length

print("Length of the longest consecutive sequence:", longest)
```

##### Question 5 -  How do you group anagrams from a list of strings?


```python
# Take input from the user
words = input("Enter words separated by spaces: ").split()

anagrams = {}

for word in words:
    key = "".join(sorted(word))   # Sort the letters

    if key in anagrams:
        anagrams[key].append(word)
    else:
        anagrams[key] = [word]

print("Grouped Anagrams:")

for group in anagrams.values():
    print(group)
```

##### Question 6 - How do you traverse a binary tree level by level?

```python
from collections import deque

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

# Function for level order traversal
def level_order(root):
    if root is None:
        return

    queue = deque([root])

    while queue:
        node = queue.popleft()
        print(node.data, end=" ")

        if node.left:
            queue.append(node.left)

        if node.right:
            queue.append(node.right)

# Creating the tree
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)
root.right.right = Node(6)

level_order(root)
```

##### Question 7 - Write a Python code to implement a binary search algorithm

```python
# Take sorted list as input
numbers = list(map(int, input("Enter sorted numbers separated by spaces: ").split()))

target = int(input("Enter the number to search: "))

low = 0
high = len(numbers) - 1

found = False

while low <= high:
    mid = (low + high) // 2

    if numbers[mid] == target:
        print("Element found at index", mid)
        found = True
        break

    elif numbers[mid] < target:
        low = mid + 1

    else:
        high = mid - 1

if not found:
    print("Element not found")
```

##### Question 8 - Create a Python function that flattens a nested list.

```python
def flatten(lst):
    result = []

    for item in lst:
        if isinstance(item, list):
            result.extend(flatten(item))
        else:
            result.append(item)

    return result


# User input
nested = eval(input("Enter a nested list: "))

print("Flattened List:", flatten(nested))
```

##### Question 9 - Write a Python program to check if a binary tree is balanced

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def height(root):
    if root is None:
        return 0

    left_height = height(root.left)
    right_height = height(root.right)

    return max(left_height, right_height) + 1


def isBalanced(root):
    if root is None:
        return True

    left_height = height(root.left)
    right_height = height(root.right)

    if abs(left_height - right_height) > 1:
        return False

    return isBalanced(root.left) and isBalanced(root.right)


# Creating a balanced tree
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)

if isBalanced(root):
    print("Tree is Balanced")
else:
    print("Tree is Not Balanced")
```

##### Question 11 - Write a Python program to count vowels and consonants in a string.

```python
text = input("Enter a string: ")
text_lower = text.lower()
vowels = set("aeiou")
vowel_count = sum(1 for ch in text_lower if ch.isalpha() and ch in vowels)
consonant_count = sum(1 for ch in text_lower if ch.isalpha() and ch not in vowels)

print("Vowels:", vowel_count)
print("Consonants:", consonant_count)
```

##### Question 13 - Write a Python program to merge two dictionaries and sum values for duplicate keys.

```python
# Example input: a:1 b:2
# Use space-separated key:value pairs

def parse_dict(prompt):
    pairs = input(prompt).split()
    result = {}
    for pair in pairs:
        key, value = pair.split(":")
        result[key] = result.get(key, 0) + int(value)
    return result

obj1 = parse_dict("Enter dictionary 1 pairs (key:value): ")
obj2 = parse_dict("Enter dictionary 2 pairs (key:value): ")

merged = obj1.copy()
for key, value in obj2.items():
    merged[key] = merged.get(key, 0) + value

print("Merged dictionary:", merged)
```

##### Question 17 - Write a Python program to validate a password with length, digit, uppercase, lowercase, and special character requirements.

```python
password = input("Enter a password: ")
length_ok = len(password) >= 8
has_digit = any(ch.isdigit() for ch in password)
has_upper = any(ch.isupper() for ch in password)
has_lower = any(ch.islower() for ch in password)
has_special = any(not ch.isalnum() for ch in password)

is_valid = all([length_ok, has_digit, has_upper, has_lower, has_special])
print("Password valid:", is_valid)
```

##### Question 18 - Write a Python program to print a histogram of character frequencies from input.

```python
text = input("Enter text: ")
frequency = {}
for ch in text:
    if ch != ' ':
        frequency[ch] = frequency.get(ch, 0) + 1

for ch in sorted(frequency):
    print(f"{ch}: {'*' * frequency[ch]}")
```

##### Question 20 - Write a Python program to implement a basic calculator using functions and user input.

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    return a / b if b != 0 else None

x = float(input("Enter first number: "))
y = float(input("Enter second number: "))
operator = input("Enter operator (+, -, *, /): ")

result = None
if operator == "+":
    result = add(x, y)
elif operator == "-":
    result = subtract(x, y)
elif operator == "*":
    result = multiply(x, y)
elif operator == "/":
    result = divide(x, y)
else:
    print("Invalid operator")

if result is not None:
    print("Result:", result)
```
